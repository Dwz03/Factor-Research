## IC matrices and complete descriptive statistics

Individual samples use each combination?s available pairs. Strict common samples intersect all factor/return pairs first, recompute ranks, then intersect valid dates. Subperiod labels are bounded independently; pooled labels can cross 2016/2017. All forward endpoints stop by 2019-12-31.

```python
for period in ['2010_2019', '2010_2016', '2017_2019']:
    for sample in ['individual', 'common']:
        rows = summary.query('period == @period and sample == @sample')
        print(period, sample)
        display(rows.pivot(index='lookback', columns='horizon', values='mean'))
        display(rows[['lookback','horizon','count','mean','std','icir','positive_fraction']])
```
